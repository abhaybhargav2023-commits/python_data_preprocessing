# Python Data Processing

## Project Overview
This project processes structured student, employee, and sales records
using Python lists, dictionaries, loops, and reusable functions.

## Objectives
- Process structured records.
- Calculate useful summary statistics.
- Generate examples of processed data.
- Visualize student ages, department-wise salaries, and monthly sales revenue.

## Tools
- Python
- Jupyter Notebook
- Matplotlib

```python
# Student records
students = [
    {"id": "S001", "name": "Aditi", "age": 20, "class": "B.Tech", "grade": "A"},
    {"id": "S002", "name": "Rohit", "age": 21, "class": "B.Sc", "grade": "B"},
    {"id": "S003", "name": "Neha", "age": 19, "class": "B.Tech", "grade": "A+"},
    {"id": "S004", "name": "Karan", "age": 22, "class": "BCA", "grade": "B"},
    {"id": "S005", "name": "Priya", "age": 20, "class": "B.Sc", "grade": "A"},
]

# Employee records
employees = [
    {"id": "E001", "name": "Amit", "department": "HR", "salary": 50000},
    {"id": "E002", "name": "Sneha", "department": "IT", "salary": 65000},
    {"id": "E003", "name": "Rahul", "department": "Finance", "salary": 55000},
    {"id": "E004", "name": "Pooja", "department": "Marketing", "salary": 60000},
    {"id": "E005", "name": "Vikram", "department": "IT", "salary": 70000},
]

# Sales records (sample revenue in rupees)
sales = [
    {"month": "Jan", "product": "Laptop", "quantity": 10, "revenue": 500000},
    {"month": "Feb", "product": "Mobile", "quantity": 15, "revenue": 450000},
    {"month": "Mar", "product": "Tablet", "quantity": 8, "revenue": 320000},
    {"month": "Apr", "product": "Laptop", "quantity": 12, "revenue": 600000},
    {"month": "May", "product": "Mobile", "quantity": 20, "revenue": 700000},
]

print("Student records:", len(students))
print("Employee records:", len(employees))
print("Sales records:", len(sales))
```

```python
def student_statistics(records):
    total_students = len(records)
    average_age = sum(student["age"] for student in records) / total_students

    grade_counts = {}
    for student in records:
        grade = student["grade"]
        grade_counts[grade] = grade_counts.get(grade, 0) + 1

    return total_students, average_age, grade_counts


total_students, average_age, grade_counts = student_statistics(students)

print("===== Student Statistics =====")
print("Total Students:", total_students)
print("Average Age:", round(average_age, 2))
print("Grade Count:", grade_counts)
```

```python
def employee_statistics(records):
    total_employees = len(records)
    average_salary = sum(employee["salary"] for employee in records) / total_employees

    department_counts = {}
    department_salary_totals = {}

    for employee in records:
        department = employee["department"]
        department_counts[department] = department_counts.get(department, 0) + 1
        department_salary_totals[department] = (
            department_salary_totals.get(department, 0) + employee["salary"]
        )

    average_salary_by_department = {
        department: department_salary_totals[department] / department_counts[department]
        for department in department_counts
    }

    return total_employees, average_salary, department_counts, average_salary_by_department


total_employees, average_salary, department_counts, department_averages = (
    employee_statistics(employees)
)

print("===== Employee Statistics =====")
print("Total Employees:", total_employees)
print("Average Salary: ₹", round(average_salary, 2))
print("Department Count:", department_counts)
print("Average Salary by Department:", department_averages)
```

```python
def sales_statistics(records):
    total_revenue = sum(record["revenue"] for record in records)
    average_revenue = total_revenue / len(records)

    highest_sale = max(records, key=lambda record: record["revenue"])
    lowest_sale = min(records, key=lambda record: record["revenue"])

    return total_revenue, average_revenue, highest_sale, lowest_sale


total_revenue, average_revenue, highest_sale, lowest_sale = sales_statistics(sales)

print("===== Sales Statistics =====")
print("Total Sales Revenue: ₹", total_revenue)
print("Average Revenue: ₹", round(average_revenue, 2))
print("Highest Revenue:", highest_sale["month"], "₹", highest_sale["revenue"])
print("Lowest Revenue:", lowest_sale["month"], "₹", lowest_sale["revenue"])
```

```python
import matplotlib.pyplot as plt

# 1. Student age distribution
ages = [student["age"] for student in students]

plt.figure(figsize=(6, 4))
plt.hist(ages, bins=[18, 19, 20, 21, 22, 23], edgecolor="black")
plt.title("Student Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.xticks([19, 20, 21, 22])
plt.tight_layout()
plt.show()

# 2. Average salary by department
departments = list(department_averages.keys())
average_salaries = list(department_averages.values())

plt.figure(figsize=(7, 4))
plt.bar(departments, average_salaries, edgecolor="black")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (₹)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# 3. Monthly sales revenue
months = [record["month"] for record in sales]
revenues = [record["revenue"] for record in sales]

plt.figure(figsize=(7, 4))
plt.plot(months, revenues, marker="o")
plt.title("Monthly Sales Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

- Student records provide an overview of student ages and grade distribution.
- Employee records help compare salary information across departments.
- Sales records help identify total revenue, average revenue, and the highest and lowest revenue months.
- Charts provide a quick visual summary of the processed data.
